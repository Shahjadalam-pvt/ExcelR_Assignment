```python
import pandas as pd
import numpy as np

import matplotlib.pyplot as plt
import seaborn as sns

# Load dataset
df = pd.read_csv("sonardataset.csv")

df.head()
```

```python
print("Dataset Shape:", df.shape)
```

```python
df.describe()
```

```python
print("Missing values:")
print(df.isnull().sum().sum())
```

```python
print("Target distribution:")
print(df.iloc[:, -1].value_counts())
```

```python
# Part 2 — Data Preprocessing & Normalization

X = df.iloc[:, :-1]
y = df.iloc[:, -1]

print("Features shape:", X.shape)
print("Target shape:", y.shape)
```

```python
print("Missing values in features:", X.isnull().sum().sum())
print("Missing values in target:", y.isnull().sum())
```

```python
y = y.map({
    "M": 1,
    "R": 0
})
```

```python
print(y.value_counts())
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("Training samples:", X_train.shape[0])
print("Testing samples:", X_test.shape[0])
```

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

X_train_scaled = scaler.fit_transform(X_train)

X_test_scaled = scaler.transform(X_test)
```

```python
print("Training mean:", X_train_scaled.mean())
print("Training standard deviation:", X_train_scaled.std())
```

```python
print("X_train:", X_train_scaled.shape)
print("X_test :", X_test_scaled.shape)
print("y_train:", y_train.shape)
print("y_test :", y_test.shape)
```

```python
# Part 3 — Build the Basic ANN

import tensorflow as tf

from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense
```

```python
print("TensorFlow version:", tf.__version__)
```

```python
model = Sequential([
    Dense(32, activation="relu", input_shape=(X_train_scaled.shape[1],)),
    Dense(16, activation="relu"),
    Dense(1, activation="sigmoid")
])
```

```python
model.compile(
    optimizer="adam",
    loss="binary_crossentropy",
    metrics=["accuracy"]
)
```

```python
model.summary()
```

```python
# Part 3 — Step 5: Train the Baseline ANN

history = model.fit(
    X_train_scaled,
    y_train,
    epochs=50,
    batch_size=16,
    validation_split=0.2,
    verbose=1
)
```

```python
plt.figure(figsize=(8, 5))

plt.plot(
    history.history["accuracy"],
    label="Training Accuracy"
)

plt.plot(
    history.history["val_accuracy"],
    label="Validation Accuracy"
)

plt.title("Baseline ANN Accuracy")
plt.xlabel("Epoch")
plt.ylabel("Accuracy")
plt.legend()
plt.grid(True)

plt.show()
```

```python
plt.figure(figsize=(8, 5))

plt.plot(
    history.history["loss"],
    label="Training Loss"
)

plt.plot(
    history.history["val_loss"],
    label="Validation Loss"
)

plt.title("Baseline ANN Loss")
plt.xlabel("Epoch")
plt.ylabel("Loss")
plt.legend()
plt.grid(True)

plt.show()
```

```python
test_loss, test_accuracy = model.evaluate(
    X_test_scaled,
    y_test,
    verbose=0
)

print("Test Loss:", test_loss)
print("Test Accuracy:", test_accuracy)
```

```python
y_pred_probability = model.predict(
    X_test_scaled
)

y_pred = (
    y_pred_probability >= 0.5
).astype(int).ravel()
```

```python
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score
)

baseline_accuracy = accuracy_score(
    y_test,
    y_pred
)

baseline_precision = precision_score(
    y_test,
    y_pred
)

baseline_recall = recall_score(
    y_test,
    y_pred
)

baseline_f1 = f1_score(
    y_test,
    y_pred
)

print("Baseline ANN Performance")
print("------------------------")
print("Accuracy :", baseline_accuracy)
print("Precision:", baseline_precision)
print("Recall   :", baseline_recall)
print("F1-Score :", baseline_f1)
```

```python
from sklearn.metrics import confusion_matrix, ConfusionMatrixDisplay

cm = confusion_matrix(
    y_test,
    y_pred
)

disp = ConfusionMatrixDisplay(
    confusion_matrix=cm,
    display_labels=["Rock", "Mine"]
)

disp.plot()

plt.title("Baseline ANN Confusion Matrix")
plt.show()
```

```python
from sklearn.metrics import classification_report

print(
    classification_report(
        y_test,
        y_pred,
        target_names=["Rock", "Mine"]
    )
)
```

## Hyperparameter Tuning Methodology

A structured grid search was used to evaluate different ANN configurations.

The following hyperparameters were varied:

- Hidden layers: 1 and 2
- Neurons per hidden layer: 16 and 32
- Activation functions: ReLU and Tanh
- Learning rates: 0.001 and 0.01

These combinations resulted in 16 different configurations. Each configuration was trained for 30 epochs using a batch size of 16 and a 20% validation split.

The F1-score was used as the primary criterion for identifying the best configuration because it provides a balance between precision and recall. Accuracy, precision, and recall were also recorded for comparison.

```python
# Part 4 — Hyperparameter Tuning

from sklearn.model_selection import ParameterGrid
```

```python
import tensorflow as tf

from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense

print("TensorFlow version:", tf.__version__)
```

```python
def build_ann(
    hidden_layers=1,
    neurons=32,
    activation="relu",
    learning_rate=0.001
):
    
    model = Sequential()
    
    # First hidden layer
    model.add(
        Dense(
            neurons,
            activation=activation,
            input_shape=(X_train_scaled.shape[1],)
        )
    )
    
    # Additional hidden layers
    for _ in range(hidden_layers - 1):
        model.add(
            Dense(
                neurons,
                activation=activation
            )
        )
    
    # Output layer
    model.add(
        Dense(
            1,
            activation="sigmoid"
        )
    )
    
    optimizer = tf.keras.optimizers.Adam(
        learning_rate=learning_rate
    )
    
    model.compile(
        optimizer=optimizer,
        loss="binary_crossentropy",
        metrics=["accuracy"]
    )
    
    return model
```

```python
param_grid = {
    "hidden_layers": [1, 2],
    "neurons": [16, 32],
    "activation": ["relu", "tanh"],
    "learning_rate": [0.001, 0.01]
}
```

```python
len(list(ParameterGrid(param_grid)))
```

```python
results = []

for params in ParameterGrid(param_grid):
    
    print("Testing:", params)
    
    tuned_model = build_ann(
        hidden_layers=params["hidden_layers"],
        neurons=params["neurons"],
        activation=params["activation"],
        learning_rate=params["learning_rate"]
    )
    
    history_tuned = tuned_model.fit(
        X_train_scaled,
        y_train,
        epochs=30,
        batch_size=16,
        validation_split=0.2,
        verbose=0
    )
    
    loss, accuracy = tuned_model.evaluate(
        X_test_scaled,
        y_test,
        verbose=0
    )
    
    y_prob = tuned_model.predict(
        X_test_scaled,
        verbose=0
    )
    
    y_pred_tuned = (
        y_prob >= 0.5
    ).astype(int).ravel()
    
    results.append({
        **params,
        "accuracy": accuracy,
        "precision": precision_score(
            y_test,
            y_pred_tuned,
            zero_division=0
        ),
        "recall": recall_score(
            y_test,
            y_pred_tuned,
            zero_division=0
        ),
        "f1_score": f1_score(
            y_test,
            y_pred_tuned,
            zero_division=0
        )
    })
```

```python
tuning_results = pd.DataFrame(results)

tuning_results.sort_values(
    by="f1_score",
    ascending=False
).head(10)
```

```python
best_result = tuning_results.loc[
    tuning_results["f1_score"].idxmax()
]

print("Best Hyperparameters:")
print(best_result)
```

```python
plt.figure(figsize=(10, 6))

sns.barplot(
    data=tuning_results.sort_values(
        "f1_score",
        ascending=False
    ).head(10),
    x="f1_score",
    y=tuning_results.sort_values(
        "f1_score",
        ascending=False
    ).head(10).index
)

plt.title("Top 10 Hyperparameter Configurations by F1-Score")
plt.xlabel("F1-Score")
plt.ylabel("Configuration Index")

plt.show()
```

```python
best_params = {
    "hidden_layers": int(best_result["hidden_layers"]),
    "neurons": int(best_result["neurons"]),
    "activation": best_result["activation"],
    "learning_rate": float(best_result["learning_rate"])
}

print(best_params)
```

```python
# Part 5 — Final Tuned ANN & Comparison

best_result = tuning_results.loc[
    tuning_results["f1_score"].idxmax()
]

print("Best Configuration:")
print(best_result)
```

```python
best_params = {
    "hidden_layers": int(best_result["hidden_layers"]),
    "neurons": int(best_result["neurons"]),
    "activation": best_result["activation"],
    "learning_rate": float(best_result["learning_rate"])
}

print(best_params)
```

```python
final_model = build_ann(
    hidden_layers=best_params["hidden_layers"],
    neurons=best_params["neurons"],
    activation=best_params["activation"],
    learning_rate=best_params["learning_rate"]
)

final_model.summary()
```

```python
final_history = final_model.fit(
    X_train_scaled,
    y_train,
    epochs=50,
    batch_size=16,
    validation_split=0.2,
    verbose=1
)
```

```python
final_loss, final_accuracy = final_model.evaluate(
    X_test_scaled,
    y_test,
    verbose=0
)

print("Final Tuned ANN Accuracy:", final_accuracy)
```

```python
final_probability = final_model.predict(
    X_test_scaled,
    verbose=0
)

final_pred = (
    final_probability >= 0.5
).astype(int).ravel()
```

```python
final_precision = precision_score(
    y_test,
    final_pred,
    zero_division=0
)

final_recall = recall_score(
    y_test,
    final_pred,
    zero_division=0
)

final_f1 = f1_score(
    y_test,
    final_pred,
    zero_division=0
)

print("Final Tuned ANN Performance")
print("---------------------------")
print("Accuracy :", final_accuracy)
print("Precision:", final_precision)
print("Recall   :", final_recall)
print("F1-Score :", final_f1)
```

Analysis and Discussion

The baseline Artificial Neural Network was first trained using a predefined architecture. Hyperparameter tuning was then performed by systematically varying the number of hidden layers, number of neurons, activation function, and learning rate.

The tuned model was selected based on its F1-score and was evaluated using the same test dataset as the baseline model. Accuracy, precision, recall, and F1-score were used to compare the two models.

The comparison shows how changing the ANN architecture and training parameters affected classification performance. The confusion matrix provides additional insight into the number of correctly and incorrectly classified Mine and Rock samples.

The training and validation curves were also examined to identify potential overfitting or underfitting. A large difference between training and validation performance can indicate overfitting, while consistently low performance on both can indicate underfitting.

Conclusion

An Artificial Neural Network was developed to classify sonar signals as Mine or Rock. The dataset contained 208 samples with 60 numerical input features. The data was explored, the target labels were encoded, and the input features were standardized before model training.

A baseline ANN was first trained using two hidden layers with 32 and 16 neurons. Hyperparameter tuning was then performed by systematically testing different numbers of hidden layers, neurons, activation functions, and learning rates.

The baseline ANN achieved an accuracy of 92.86%, precision of 91.30%, recall of 95.45%, and F1-score of 93.33%. The final tuned ANN achieved an accuracy of 90.48%, precision of 87.50%, recall of 95.45%, and F1-score of 91.30%.

Therefore, for the final training run, the baseline ANN performed better overall than the tuned ANN based on F1-score, accuracy, and precision. Although the tuning search identified a configuration with a higher F1-score during the search process, retraining the selected configuration produced different results because neural network training involves random initialization and stochastic optimization.

The results demonstrate that hyperparameter tuning does not necessarily guarantee better performance on a particular train/test split. Since the dataset is relatively small, model performance can vary between different splits and training runs.
